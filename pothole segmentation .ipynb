```python
import cv2
import os
from ultralytics import YOLO
from IPython.display import Video

# Step 1: Create output folder if not exists
os.makedirs('output', exist_ok=True)

# Step 2: Load your YOLOv8 segmentation model
model = YOLO('best pothole segmentation.pt')  # Make sure this file is in the same directory

# Step 3: Load your input video
cap = cv2.VideoCapture('sample_video pothole.mp4')  # Replace with your actual video file

# Step 4: Get video details
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Step 5: Define output video file name (no spaces!)
output_path = "output/pothole_segmentation_1.mp4"
out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

# Step 6: Process video frame by frame
while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)
    annotated_frame = results[0].plot()
    out.write(annotated_frame)

# Step 7: Release everything
cap.release()
out.release()

print(f"✅ Video saved successfully at {output_path}")

# Step 8: Display the output video in Jupyter Notebook
Video(output_path, width=640, height=480)
```
